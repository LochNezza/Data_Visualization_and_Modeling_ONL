## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed.
```python
critic_genre_df = video_games_df.dropna(
    subset=["Genre", "Critic_Score"]
)

critic_genre_df[["Genre", "Critic_Score"]].head()
```

```python
critic_genre_df = critic_genre_df.copy()

critic_genre_df["Score_80_Plus"] = critic_genre_df["Critic_Score"] >= 80

critic_genre_df[["Genre", "Critic_Score", "Score_80_Plus"]].head()
```

```python
critic_genre_df.groupby("Genre")["Score_80_Plus"].mean().mul(100).sort_values(ascending=False)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:


I created a Boolean feature called Score_80_Plus that identifies whether each game with a recorded critic score received a Critic_Score of 80 or higher. Sports had the highest percentage of games scoring 80 or higher at approximately 35.09%, followed by Role-Playing at 31.61% and Shooter at 29.24%. Puzzle had the lowest percentage at approximately 15.18%. These calculations show that the percentage of games receiving a critic score of 80 or higher varies considerably among genres.
